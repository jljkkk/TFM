```python
import requests
import pandas as pd
import io

def descargar_zona_cuadricula(fecha_inicio, fecha_fin, mag_min, min_lat, max_lat, min_lon, max_lon):
    url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
    params = {
        "format": "csv",
        "starttime": fecha_inicio,
        "endtime": fecha_fin,
        "minmagnitude": mag_min,
        # Filtros geográficos:
        "minlatitude": min_lat,
        "maxlatitude": max_lat,
        "minlongitude": min_lon,
        "maxlongitude": max_lon
    }
    
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return pd.read_csv(io.StringIO(response.text))
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python
import time
import pandas as pd

# Definimos las coordenadas de la cuadrícula de Japón
min_lat_japon = 24.0
max_lat_japon = 46.0
min_lon_japon = 122.0
max_lon_japon = 146.0

fechas = pd.date_range(start="2020-01-01", end="2024-01-01", freq="MS")
lista_df = []

print("Iniciando extracción mes a mes para Japón...")

for i in range(len(fechas) - 1):
    inicio = fechas[i].strftime("%Y-%m-%d")
    fin = fechas[i+1].strftime("%Y-%m-%d")
    
    print(f"Descargando {inicio} a {fin}...", end=" ")
    
    df_mes = descargar_zona_cuadricula(
        inicio, fin, 4.0,  # Nota: Subido a 4.0 (leer consejo abajo)
        min_lat_japon, max_lat_japon, 
        min_lon_japon, max_lon_japon
    )
    
    if df_mes is not None:
        print(f"{len(df_mes)} sismos.")
        lista_df.append(df_mes)
        
    time.sleep(0.5)

if lista_df:
    df_japon_total = pd.concat(lista_df, ignore_index=True)
    df_japon_total = df_japon_total.drop_duplicates(subset='id')
    df_japon_total = df_japon_total.sort_values(by='time').reset_index(drop=True)
    
    print(f"\n¡Éxito Total! Se han descargado y unificado {len(df_japon_total)} sismos de Japón.")
    df_japon_total.to_csv("catalogo_japon_2020_2023.csv", index=False)
```

```python
df_japon['magType'] = df_japon['magType'].astype(str).str.lower()

# 1.2 Agrupar subtipos a sus clases principales
diccionario_reemplazo = {
    'mww': 'mw', 'mwr': 'mw',  # Derivados de Magnitud de Momento
    'mlr': 'ml',               # Derivado de Magnitud Local
    'mc': 'md'                 # Coda duration -> Magnitud de Duración
}
df_japon['magType'] = df_japon['magType'].replace(diccionario_reemplazo)

df_chile['magType'] = df_chile['magType'].astype(str).str.lower()

# 1.2 Agrupar subtipos a sus clases principales
diccionario_reemplazo = {
    'mww': 'mw', 'mwr': 'mw',  # Derivados de Magnitud de Momento
    'mlr': 'ml',               # Derivado de Magnitud Local
    'mc': 'md'                 # Coda duration -> Magnitud de Duración
}
df_chile['magType'] = df_chile['magType'].replace(diccionario_reemplazo)
```

```python
df_japon= df_japon[~df_japon['magType'].isin(["mh","ma"])]
df_japon["magType"].value_counts()

df_chile= df_chile[~df_chile['magType'].isin(["mh","ma"])]
df_chile["magType"].value_counts()
```

```python
import numpy as np
def convert_to_mw(mag, mag_type):
    if mag_type == "mw":
        return mag
    elif mag_type == "mb":
        return 1.085 * mag - 0.45
    elif mag_type == "md":
        return 1.193 * mag - 0.52
    elif mag_type == "ml":
        return 0.906 * mag + 0.02
    else:
        return np.nan 

```

```python
df_japon["mag_Mw"] = df_japon.apply(lambda row: convert_to_mw(row['mag'], row['magType']), axis=1)
df_chile["mag_Mw"] = df_chile.apply(lambda row: convert_to_mw(row['mag'], row['magType']), axis=1)
```

```python
df_japon = df_japon.drop(columns=['mag'])
df_chile = df_chile.drop(columns=['mag'])
```

```python
df_japon['energia_julios'] = 10 ** (1.44 * df_japon['mag_Mw'] + 5.24)
df_chile['energia_julios'] = 10 ** (1.44 * df_chile['mag_Mw'] + 5.24)
```

```python
# ==========================================
# 1. CREACIÓN DE LA CUADRÍCULA ESPACIAL (GRID)
# ==========================================
# Redondeamos lat/lon para crear celdas (Ej: 1 grado x 1 grado)
df_japon['lat_bin'] = df_japon['latitude'].round(0)
df_japon['lon_bin'] = df_japon['longitude'].round(0)
df_japon['cell_id'] = df_japon['lat_bin'].astype(str) + "_" + df_japon['lon_bin'].astype(str)
df_japon["cell_id"].value_counts()

df_chile['lat_bin'] = df_chile['latitude'].round(0)
df_chile['lon_bin'] = df_chile['longitude'].round(0)
df_chile['cell_id'] = df_chile['lat_bin'].astype(str) + "_" + df_chile['lon_bin'].astype(str)
df_chile["cell_id"].value_counts()
```

```python
df_japon['time'] = pd.to_datetime(df_japon['time'])
df_japon['date'] = df_japon['time'].dt.floor('D')
df_chile['time'] = pd.to_datetime(df_chile['time'])
df_chile['date'] = df_chile['time'].dt.floor('D')
```

```python
# 1. Calculamos las estadísticas generales (como ya hacías)
daily_stats_japon = df_japon.groupby(['cell_id', 'date']).agg(
    eq_count=('mag_Mw', 'count'),
    energia_acumulada_diaria=('energia_julios', 'sum')
).reset_index()

# 2. Identificamos la fila del sismo más fuerte para cada celda y día
idx_max = df_japon.groupby(['cell_id', 'date'])['mag_Mw'].idxmax()
main_events = df_japon.loc[idx_max].copy()

# 3. Renombramos las columnas del sismo principal para no confundirlas
# Por ejemplo: 'depth' pasa a ser 'main_depth'
main_events = main_events.rename(columns={
    'mag_Mw': 'main_mag',
    'depth': 'main_depth',
    'latitude': 'main_lat',
    'longitude': 'main_lon'
})

# 4. Unimos todo en un solo DataFrame de "Panel"
# Ahora cada día tiene: cuántos sismos hubo Y los detalles del sismo más fuerte
final_daily_japon = pd.merge(
    daily_stats_japon, 
    main_events[['cell_id', 'date', 'main_mag', 'main_depth', 'main_lat', 'main_lon',
                 'magError', 'depthError', 'horizontalError', 'nst', 'gap', 'dmin', 'rms', 'magNst'
                 , 'lat_bin', 'lon_bin', 'locationSource', 'magSource',
                 'net', 'status', 'updated']], 
    on=['cell_id', 'date'], 
    how='left'
)

# 1. Calculamos las estadísticas generales para Chile
daily_stats_chile = df_chile.groupby(['cell_id', 'date']).agg(
    eq_count=('mag_Mw', 'count'),
    energia_acumulada_diaria=('energia_julios', 'sum')
).reset_index()
# 2. Identificamos la fila del sismo más fuerte para cada celda y día
idx_max = df_chile.groupby(['cell_id', 'date'])['mag_Mw'].idxmax()
main_events = df_chile.loc[idx_max].copy()

# 3. Renombramos las columnas del sismo principal para no confundirlas
idx_max = df_chile.groupby(['cell_id', 'date'])['mag_Mw'].idxmax()
idx_max = (
    df_chile.dropna(subset=['mag_Mw'])
    .groupby(['cell_id', 'date'])['mag_Mw']
    .idxmax()
    .dropna()
    .astype(int)
)

main_events = df_chile.loc[idx_max].copy()

# 3. Renombramos las columnas del sismo principal para no confundirlas
# Por ejemplo: 'depth' pasa a ser 'main_depth'
main_events = main_events.rename(columns={
    'mag_Mw': 'main_mag',
    'depth': 'main_depth',
    'latitude': 'main_lat',
    'longitude': 'main_lon'
})

# 4. Unimos todo en un solo DataFrame de "Panel"
# Ahora cada día tiene: cuántos sismos hubo Y los detalles del sismo más fuerte
final_daily_chile = pd.merge(
    daily_stats_chile, 
    main_events[['cell_id', 'date', 'main_mag', 'main_depth', 'main_lat', 'main_lon',
                 'magError', 'depthError', 'horizontalError', 'nst', 'gap', 'dmin', 'rms', 'magNst'
                 , 'lat_bin', 'lon_bin', 'locationSource', 'magSource',
                 'net', 'status', 'updated']], 
    on=['cell_id', 'date'], 
    how='left'
)
```

```python
# ==========================================
# 4. RELLENAR LOS DÍAS VACÍOS (ZEROS)
# ==========================================
# Creamos un MultiIndex con todas las combinaciones posibles de Fechas y Celdas
all_dates = pd.date_range(start=daily_stats_japon['date'].min(), end=daily_stats_japon['date'].max(), freq='D')
all_dates = pd.to_datetime(all_dates).date
all_cells = daily_stats_japon['cell_id'].unique()
full_index = pd.MultiIndex.from_product([all_cells, all_dates], names=['cell_id', 'date'])

# Reindexamos y rellenamos los nulos (días sin sismos) con ceros
panel_df_japon = final_daily_japon.set_index(['cell_id', 'date']).reindex(full_index).reset_index()

all_dates = pd.date_range(start=daily_stats_chile['date'].min(), end=daily_stats_chile['date'].max(), freq='D')
all_dates = pd.to_datetime(all_dates).date
all_cells = daily_stats_chile['cell_id'].unique()
full_index = pd.MultiIndex.from_product([all_cells, all_dates], names=['cell_id', 'date'])

panel_df_chile = final_daily_chile.set_index(['cell_id', 'date']).reindex(full_index).reset_index()
```

```python
offsets = [
    (-1, -1), (-1, 0), (-1, 1), # Sur-Oeste, Sur, Sur-Este
    ( 0, -1),          ( 0, 1), # Oeste,          Este
    ( 1, -1), ( 1, 0), ( 1, 1)  # Nor-Oeste, Norte, Nor-Este
]

shifted_dfs = []

# Nombres de tus columnas originales (CÁMBIALOS SI SE LLAMAN DISTINTO EN TU DF)
col_fecha = 'date'
col_n_sismos = 'eq_count'
col_mag = 'main_mag'
col_roll7_eq_count = 'roll7_eq_count'
col_roll7_mag_max = 'roll7_mag_max'
col_roll7_energia = 'roll7_energia'
col_roll30_energia = 'roll30_energia'
col_roll30_eq_count = 'roll30_eq_count'
col_roll30_mag_max = 'roll30_mag_max'

for d_lat, d_lon in offsets:
    # Hacemos una copia solo de las columnas necesarias
    df_neighbor_j = df_j[[col_fecha, 'lat', 'lon', col_n_sismos, col_mag, col_roll7_eq_count, col_roll7_mag_max, col_roll30_energia, col_roll30_eq_count, col_roll30_mag_max]].copy()
    
    # TRUCO MÁGICO: "Movemos" las coordenadas de esta celda hacia su vecina.
    # Al restar el offset, los datos de la celda vecina caerán matemáticamente
    # encima de las coordenadas de nuestra celda objetivo.
    df_neighbor_j['lat'] = df_neighbor_j['lat'] - d_lat
    df_neighbor_j['lon'] = df_neighbor_j['lon'] - d_lon
    
    # Renombrar para saber que son datos del vecino
    df_neighbor_j = df_neighbor_j.rename(columns={
        col_n_sismos: 'neigh_num_eqs',
        col_mag: 'neigh_mag',
        col_roll7_eq_count: 'neigh_roll7_eq_count',
        col_roll7_mag_max: 'neigh_roll7_mag_max',
        col_roll30_energia: 'neigh_roll30_energia',
        col_roll30_eq_count: 'neigh_roll30_eq_count',
        col_roll30_mag_max: 'neigh_roll30_mag_max'
    })
    
    shifted_dfs.append(df_neighbor_j)

all_neighbors = pd.concat(shifted_dfs, ignore_index=True)

# 5. Agrupar por Fecha, Latitud y Longitud objetivo para calcular el total de los 8 vecinos
vecinos_agrupados = all_neighbors.groupby([col_fecha, 'lat', 'lon']).agg(
    adj_num_eqs=('neigh_num_eqs', 'sum'), # SUMAR la cantidad de terremotos de los vecinos
    adj_max_mag=('neigh_mag', 'max'),     # Extraer la MAGNITUD MÁXIMA de los vecinos
    adj_roll7_eq_count=('neigh_roll7_eq_count', 'sum'), # SUMAR la cantidad de terremotos de los vecinos (ventana 7 días)
    adj_roll7_mag_max=('neigh_roll7_mag_max', 'max'),      # Extraer la MAGNITUD MÁXIMA de los vecinos (ventana 7 días)
    adj_roll30_energia=('neigh_roll30_energia', 'sum'), # SUMAR la cantidad de terremotos de los vecinos (ventana 30 días)
    adj_roll30_eq_count=('neigh_roll30_eq_count', 'sum'), # SUMAR la cantidad de terremotos de los vecinos (ventana 30 días)
    adj_roll30_mag_max=('neigh_roll30_mag_max', 'max')      # Extraer la MAGNITUD MÁXIMA de los vecinos (ventana 30 días)

).reset_index()

# 6. Unimos estas nuevas variables de vuelta a nuestro DataFrame original
df_j = df_j.merge(
    vecinos_agrupados, 
    on=[col_fecha, 'lat', 'lon'], 
    how='left'
)

```
